## Evolution Strategy (μ + λ) — Function Reference

This section implements a **(μ + λ) Evolution Strategy (ES)** for the Job Shop Scheduling Problem (JSSP) using a **permutation-with-repetition** chromosome encoding (each job ID appears `num_machines` times).

---

## `read_jobs(filename)`
Reads a JSSP instance from a `.txt` file.

- The first line contains: `num_jobs num_machines`
- Each subsequent line describes one job as pairs: `(machine, duration)`

**Returns:** `jobs`  
Where `jobs[job_id]` is a list of operations `(machine, duration)` in processing order.

---

## `random_chrom(num_jobs, num_machines)`
Creates a random chromosome (schedule encoding).

- Chromosome length: `num_jobs * num_machines`
- Each job ID appears exactly `num_machines` times
- The list is randomly shuffled

**Returns:** `chrom` (list of ints)

**Why it is valid:** the decoder (`makespan`) interprets each occurrence of a job as “schedule the next operation of that job”.

---

## `makespan(jobs, chrom)`
Evaluates a chromosome by simulating the schedule and returning the **makespan** (Cmax).

**How it works (decoder):**
- `job_op[j]` tracks which operation index is next for job `j`
- `job_time[j]` tracks when job `j` becomes available
- `machine_time[m]` tracks when machine `m` becomes available
- For each gene `job` in `chrom`, schedule job’s next operation on its required machine at:
  `start = max(job_time[job], machine_time[machine])`

**Returns:** `Cmax` (int) = the maximum completion time over all machines.

---

## `mutate(chrom)`
Mutation operator used to generate offspring from a parent chromosome.

**Current implementation:** *swap mutation*
- copies the chromosome
- chooses two random indices `i, j`
- swaps `chrom[i]` and `chrom[j]`

**Returns:** mutated chromosome (list of ints)

**Property:** preserves feasibility because it only reorders genes (job counts remain unchanged).

---

## `run(filename, mu=20, lam=100, generations=500, seed=42, stop_if_stagnant=...)`
Runs one ES trial on one instance and prints progress.

**Algorithm outline:**
1. Seed RNG and read instance
2. Create `mu` random parents
3. Repeat for each generation:
   - create `lam` offspring by selecting a random parent and applying `mutate`
   - combine parents + offspring
   - select the best `mu` by lowest `makespan` (**(μ + λ) selection**)
   - track global best
   - optional early stopping if no improvement for `stop_if_stagnant` generations

**Returns:** `(best_chrom, best_makespan)`

---

## `run_es_collect_curve(file_path, mu, lam, generations, runs=20)`
Runs the ES multiple times and collects convergence statistics.

**What it collects per run:**
- `best_curve[g]`: best makespan in generation `g`
- runtime per run
- final makespan per run

**Returns:**
- `avg_curve`: average best-makespan curve over all runs
- `avg_time_per_run`: mean runtime
- `run_times`: list of runtimes
- `final_makespans`: list of final makespans (one per run)

---

## Parameter sets (`es1`, `es2`, `es3`)
Different population sizes tested:

- `es1`: μ=20, λ=100
- `es2`: μ=30, λ=150
- `es3`: μ=40, λ=200

These trade off **runtime** (more offspring evaluated) vs **solution quality / reliability** (more sampling per generation).

---

## Notes / Interpretation
- The ES tends to improve quickly early on, then plateau (diminishing returns).
- On harder instances, swap-only mutation may converge to local optima; stronger or mixed mutation operators (e.g., insert/reverse) can help escape plateaus.

## Read input file

```python
def read_jobs(filename):
    with open(filename) as f:
        lines = f.read().split("\n")
    lines = [l.strip() for l in lines if l.strip()]
    num_jobs, num_machines = map(int, lines[0].split())
    jobs = []
    for i in range(1, num_jobs + 1):
        nums = list(map(int, lines[i].split()))
        ops = [(nums[j], nums[j+1]) for j in range(0, len(nums), 2)]
        jobs.append(ops)
    return jobs
```

## Fitness function — makespan (Cmax)

```python
def makespan(jobs, chrom):
    job_op = [0] * len(jobs)
    job_time = [0] * len(jobs)
    machine_time = [0] * len(jobs[0])
    for job in chrom:
        m, d = jobs[job][job_op[job]]
        start = max(job_time[job], machine_time[m])
        end = start + d
        job_time[job] = end
        machine_time[m] = end
        job_op[job] += 1
    return max(machine_time)
```

## Chromosome initialisation

```python
def random_chrom(num_jobs, num_machines):
    c = [j for j in range(num_jobs) for _ in range(num_machines)]
    random.shuffle(c)
    return c
```

## Mutation (swap)

```python
def mutate(chrom):
    c = chrom[:]
    i, j = random.sample(range(len(c)), 2)
    c[i], c[j] = c[j], c[i]
    return c
```

## (μ + λ) main loop

```python
def run(filename, mu=20, lam=100, generations=500, seed=42, stop_if_stagnant=1340):
    random.seed(seed)
    jobs = read_jobs(filename)
    num_jobs, num_machines = len(jobs), len(jobs[0])

    # Initialise mu parents
    parents = sorted([random_chrom(num_jobs, num_machines) for _ in range(mu)],
                     key=lambda c: makespan(jobs, c))
    best = parents[0][:]
    best_ms = makespan(jobs, best)
    print(f"Gen 0 | Makespan: {best_ms}")

    stop_no_improve = 0
    stop_last_best = best_ms

    for gen in range(1, generations + 1):

        # Generate lambda offspring (1 parent → 1 mutation → 1 child) we do this for our my + lambda
        offspring = []
        for _ in range(lam):
            parent = random.choice(parents)
            child = mutate(parent)
            offspring.append(child)

        # (mu + lambda): select best mu from parents + offspring
        parents = sorted(parents + offspring, key=lambda c: makespan(jobs, c))[:mu]

        ms = makespan(jobs, parents[0])
        if ms < best_ms:
            best_ms = ms
            best = parents[0][:]

        # Early stopping check
        if ms < stop_last_best:
            stop_last_best = ms
            stop_no_improve = 0
        else:
            stop_no_improve += 1
            if stop_no_improve >= stop_if_stagnant:
                print(f"Stopping early at gen {gen} (no improvement for {stop_if_stagnant} gens)")
                break

        if gen % 50 == 0:
            print(f"Gen {gen} | Makespan: {best_ms}")

    print(f"Best makespan: {best_ms}")
    return best, best_ms
```

# Hill Climbing — Function Reference (Expanded)

## `parse_instance(file_path)`
Reads a JSSP `.txt` instance and parses the problem data.  
It finds the `jobs machines` line, then reads each job’s operations as `(machine, duration)` pairs.  
**Returns:** `(jobs, machines, jobs_data)`.

**Example (conceptual):**  
If the file starts with `3 2`, it returns 3 jobs, 2 machines, and a list of 3 jobs, each with 2 operations.

---

## `makespan(chromosome, jobs, n_machines)`
Simulates the schedule encoded by the chromosome.  
Each gene is a job ID; the function schedules that job’s **next operation** when both the job and its machine are free.  
It tracks job and machine end times and returns the **final completion time** (makespan) and a per‑machine schedule.

**Example (conceptual):**  
If the chromosome starts `[0,1,0,1]`, it schedules job 0’s first op, then job 1’s first op, then job 0’s second op, and so on, respecting machine availability.

---

## `random_chromosome(n_jobs, n_machines)`
Creates a random chromosome where each job appears exactly `n_machines` times.  
This ensures the chromosome length is `n_jobs * n_machines`.

**Example:**  
`n_jobs=3, n_machines=2` → a shuffled list like `[1,0,2,1,0,2]`.

---

## `swap_adjacent(chrom)`
Neighbourhood operator that swaps two adjacent genes at a random position.  
This makes a small, local change to the schedule.

**Example:**  
`[0,1,2,0]` → swap at index 1 gives `[0,2,1,0]`.

---

## `insert_move(chrom)`
Neighbourhood operator that removes one gene and inserts it at another index.  
This can produce a larger rearrangement than a swap.

**Example:**  
`[0,1,2,0]` → move index 0 to index 2 gives `[1,2,0,0]`.

---

## `reverse_segment(chrom)`
Neighbourhood operator that reverses a random segment of the chromosome.  
This changes the relative order of several operations at once.

**Example:**  
`[0,1,2,0,1]` → reverse positions 1–3 gives `[0,0,2,1,1]`.

---

## `hill_climbing(jobs, n_jobs, n_machines, max_iter=..., operator=None)`
Core hill‑climbing loop.  
Starts from a random chromosome and repeatedly accepts a neighbour if it is **not worse** (lower or equal makespan).  
Returns the best chromosome found, its makespan, and the convergence history.

**Example:**  
Over iterations, the makespan typically drops quickly and then plateaus.

---

## `random_restart_hill_climbing(jobs, n_jobs, n_machines, max_iter=..., n_restarts=...)`
Runs hill climbing multiple times from different random starts.  
Keeps the best solution across all restarts and concatenates histories.

**Example:**  
If one restart gets stuck in a local optimum, another restart can find a better one.

---

## `run_hc_suite(file_path, label)`
End‑to‑end experiment runner.  
It runs **Single HC** and **Random‑Restart HC**, plus per‑operator HC, and plots convergence curves.  
It also prints summary stats and shows a distribution of final makespans.

**Example:**  
Produces two convergence plots and a histogram for each instance.

```python
import numpy as np 
import matplotlib.pyplot as plt
import re
from collections import defaultdict
import time
import random

```

```python
def parse_instance(file_path):
    with open(file_path, 'r') as f:
        lines = f.readlines()

    lines = [line.strip() for line in lines if line.strip()]

    for i, line in enumerate(lines):
        if line[0].isdigit():
            jobs, machines = map(int, line.split())
            start_index = i + 1
            break

    jobs_data = []
    for line in lines[start_index:start_index + jobs]:
        numbers = list(map(int, line.split()))
        operations = []
        for i in range(0, len(numbers), 2):
            machine = numbers[i]
            duration = numbers[i + 1]
            operations.append((machine, duration))
        jobs_data.append(operations)

    return jobs, machines, jobs_data
```

```python
def makespan(chromosome, jobs, n_machines):
    n_jobs      = len(jobs)
    job_step    = [0] * n_jobs
    job_end     = [0] * n_jobs
    machine_end = [0] * n_machines
    schedule    = defaultdict(list)

    for job in chromosome:
        step = job_step[job]
        if step >= len(jobs[job]):
            continue
        machine, duration = jobs[job][step]
        start  = max(job_end[job], machine_end[machine])
        finish = start + duration
        schedule[machine].append((job, start, finish))
        machine_end[machine] = finish
        job_end[job]         = finish
        job_step[job]       += 1

    return max(machine_end), dict(schedule)


def random_chromosome(n_jobs, n_machines):
    chrom = [j for j in range(n_jobs) for _ in range(n_machines)]
    np.random.shuffle(chrom)
    return chrom
```

```python
def swap_adjacent(chrom):
    c = chrom[:]
    i = random.randrange(len(c) - 1)
    c[i], c[i+1] = c[i+1], c[i]
    return c


def insert_move(chrom):
    c = chrom[:]
    i = random.randrange(len(c))
    j = random.randrange(len(c))
    elem = c.pop(i)
    c.insert(j, elem)
    return c


def reverse_segment(chrom):
    c = chrom[:]
    i, j = sorted(random.sample(range(len(c)), 2))
    c[i:j+1] = c[i:j+1][::-1]
    return c


OPERATORS = [swap_adjacent, insert_move, reverse_segment]
```

```python
def hill_climbing(jobs, n_jobs, n_machines,
                  max_iter=10_000,
                  operator=None):

    current    = random_chromosome(n_jobs, n_machines)
    current_ms = makespan(current, jobs, n_machines)[0]
    history    = [current_ms]

    for _ in range(max_iter):
        op        = operator if operator else random.choice(OPERATORS)
        neighbour = op(current)
        nb_ms     = makespan(neighbour, jobs, n_machines)[0]

        if nb_ms <= current_ms:
            current    = neighbour
            current_ms = nb_ms

        history.append(current_ms)

    return current, current_ms, history


def random_restart_hill_climbing(jobs, n_jobs, n_machines,
                                  max_iter=10_000,
                                  n_restarts=10):

    iters_per_restart = max(1, max_iter // n_restarts)
    best_chrom   = None
    best_ms      = float('inf')
    full_history = []

    for restart in range(n_restarts):
        chrom, ms, hist = hill_climbing(
            jobs, n_jobs, n_machines,
            max_iter=iters_per_restart
        )
        full_history.extend(hist)

        if ms < best_ms:
            best_ms    = ms
            best_chrom = chrom[:]

    return best_chrom, best_ms, full_history
```

```python
KNOWN_OPTIMAL = {
    'ft06': 55, 'ft10': 930, 'ft20': 1165,
    'la01': 666, 'la02': 655, 'la03': 597,
    'abz5': 1234, 'abz6': 943, 'abz7': 656,  # add if known
}

MAX_ITER = 30_000
RESTARTS = 20
N_RUNS  = 20
HC_RR_RUNS = 20

def run_hc_suite(file_path, label):
    n_jobs, n_machines, jobs = parse_instance(file_path)
    opt = KNOWN_OPTIMAL.get(label)

    print(f"\n=== Instance {label} ===")
    print(f"Jobs: {n_jobs} | Machines: {n_machines} | Optimal: {opt}")

    # 1) Single HC (averaged over HC_RR_RUNS)
    hc_histories = []
    hc_makespans = []
    hc_times = []
    for _ in range(HC_RR_RUNS):
        t0 = time.time()
        _, hc_ms, hc_hist = hill_climbing(jobs, n_jobs, n_machines, max_iter=MAX_ITER)
        hc_times.append(time.time() - t0)
        hc_makespans.append(hc_ms)
        hc_histories.append(hc_hist)

    hc_best = min(hc_makespans)
    hc_avg = np.mean(hc_makespans)
    hc_time_avg = np.mean(hc_times)
    hc_hist_avg = [np.mean(vals) for vals in zip(*hc_histories)]

    print(f"Single HC over {HC_RR_RUNS} runs: best = {hc_best:6d} | avg = {hc_avg:6.1f} | time/run = {hc_time_avg:.2f}s")

    # 2) Random-Restart HC (averaged over HC_RR_RUNS)
    rr_histories = []
    rr_makespans = []
    rr_times = []
    for _ in range(HC_RR_RUNS):
        t0 = time.time()
        _, rr_ms, rr_hist = random_restart_hill_climbing(
            jobs, n_jobs, n_machines, max_iter=MAX_ITER, n_restarts=RESTARTS
        )
        rr_times.append(time.time() - t0)
        rr_makespans.append(rr_ms)
        rr_histories.append(rr_hist)

    rr_best = min(rr_makespans)
    rr_avg = np.mean(rr_makespans)
    rr_time_avg = np.mean(rr_times)
    rr_hist_avg = [np.mean(vals) for vals in zip(*rr_histories)]

    print(f"Random-Restart HC over {HC_RR_RUNS} runs: best = {rr_best:6d} | avg = {rr_avg:6.1f} | time/run = {rr_time_avg:.2f}s")

    # 3) Per-operator HC
    op_results = {}
    for op in OPERATORS:
        op_run_results = []
        op_hist_sample = None
        op_times = []  # ADDED

        for _ in range(N_RUNS):
            t0 = time.time()  # ADDED
            _, op_ms, op_hist = hill_climbing(
                jobs, n_jobs, n_machines, max_iter=MAX_ITER, operator=op
            )
            op_times.append(time.time() - t0)  # ADDED
            op_run_results.append(op_ms)
            if op_hist_sample is None:
                op_hist_sample = op_hist

        op_best = min(op_run_results)
        op_avg = np.mean(op_run_results)
        op_time_avg = np.mean(op_times)  # ADDED

        op_results[op.__name__] = (op_best, op_avg, op_time_avg, op_hist_sample)  # CHANGED
        print(f"HC ({op.__name__:<20}) best = {op_best:6d} | avg = {op_avg:6.1f} | time/run = {op_time_avg:.2f}s")  # CHANGED


    # Plot convergence
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    fig.suptitle(f"Hill Climbing Convergence — {label}", fontsize=13, fontweight='bold')

    ax = axes[0]
    step = max(1, len(hc_hist_avg) // 500)
    ax.plot(range(0, len(hc_hist_avg), step),
            [hc_hist_avg[i] for i in range(0, len(hc_hist_avg), step)],
            color='steelblue', lw=1.5,
            label=f'Single HC (best={hc_best}, avg={hc_avg:.1f}, time/run={hc_time_avg:.2f}s)')  # CHANGED
    step2 = max(1, len(rr_hist_avg) // 500)
    ax.plot(range(0, len(rr_hist_avg), step2),
            [rr_hist_avg[i] for i in range(0, len(rr_hist_avg), step2)],
            color='darkorange', lw=1.5,
            label=f'RR-HC x{RESTARTS} (best={rr_best}, avg={rr_avg:.1f}, time/run={rr_time_avg:.2f}s)')  # CHANGED
    ax.set_xlabel("Iteration")
    ax.set_ylabel("Makespan")
    ax.set_title("Single HC vs Random-Restart HC")
    ax.legend(fontsize=9)

    ax = axes[1]
    colors = ['steelblue', 'darkorange', 'seagreen']
    for (name, (best_ms, avg_ms, time_avg, hist)), color in zip(op_results.items(), colors):  # CHANGED
        step = max(1, len(hist) // 500)
        ax.plot(range(0, len(hist), step),
                [hist[i] for i in range(0, len(hist), step)],
                color=color, lw=1.5,
                label=f'{name} (best={best_ms}, avg={avg_ms:.1f}, time/run={time_avg:.2f}s)')  # CHANGED
    ax.set_xlabel("Iteration")
    ax.set_ylabel("Makespan")
    ax.set_title("HC per Neighbourhood Operator")
    ax.legend(fontsize=9)

    plt.tight_layout()
    plt.show()

    # Distribution over runs
    run_results = []
    for _ in range(N_RUNS):
        _, ms, _ = hill_climbing(jobs, n_jobs, n_machines, max_iter=MAX_ITER)
        run_results.append(ms)

    print(f"Over {N_RUNS} runs:")
    print(f"  Best  : {min(run_results)}")
    print(f"  Worst : {max(run_results)}")
    print(f"  Mean  : {np.mean(run_results):.1f}")
    print(f"  Std   : {np.std(run_results):.1f}")
    if opt:
        gaps = [(r - opt) / opt * 100 for r in run_results]
        print(f"  Avg gap from optimal: {np.mean(gaps):.1f}%")

    fig, ax = plt.subplots(figsize=(9, 4))
    ax.hist(run_results, bins=10, color='steelblue', edgecolor='white')
    ax.axvline(np.mean(run_results), color='darkorange', ls='--', lw=2,
               label=f'Mean ({np.mean(run_results):.0f})')
    ax.set_xlabel("Final Makespan")
    ax.set_ylabel("Count")
    ax.set_title(f"Distribution of HC Results over {N_RUNS} Runs — {label}")
    ax.legend()
    plt.tight_layout()
    plt.show()


instances = [
    ("abz5.txt", "abz5"),
    ("abz6.txt", "abz6"),
    ("abz7.txt", "abz7"),
]

for path, label in instances:
    run_hc_suite(path, label)
```

```python


OPTIMAL = {'abz5': 1234, 'abz6': 943, 'abz7': 656}

# (μ+λ) Evolution Strategy — best and avg over 20 runs
ES_RESULTS = {
    'abz5': {
        'es1 (μ=20,λ=100)': {'best': 1264, 'avg': 1310, 'time': 0.78},
        'es2 (μ=30,λ=150)': {'best': 1261, 'avg': 1308, 'time': 1.18},
        'es3 (μ=40,λ=200)': {'best': 1242, 'avg': 1297, 'time': 1.66},
    },
    'abz6': {
        'es1 (μ=20,λ=100)': {'best': 948,  'avg': 1006, 'time': 0.84},
        'es2 (μ=30,λ=150)': {'best': 961,  'avg': 1003, 'time': 1.19},
        'es3 (μ=40,λ=200)': {'best': 951,  'avg':  996, 'time': 1.58},
    },
    'abz7': {
        'es1 (μ=20,λ=100)': {'best': 757,  'avg': 789,  'time': 2.14},
        'es2 (μ=30,λ=150)': {'best': 726,  'avg': 780,  'time': 3.18},
        'es3 (μ=40,λ=200)': {'best': 734,  'avg': 767,  'time': 4.40},
    },
}

# Hill Climbing — best and avg over 20 runs
HC_RESULTS = {
    'abz5': {
        'HC single':        {'best': 1242, 'avg': 1269, 'time': 0.75},
        'HC insert_move':   {'best': 1239, 'avg': 1262, 'time': 0.69},
        'HC reverse_seg':   {'best': 1252, 'avg': 1290, 'time': 0.75},
        'HC rand-restart':  {'best': 1262, 'avg': 1288, 'time': 0.73},
    },
    'abz6': {
        'HC single':        {'best': 947,  'avg':  968, 'time': 0.73},
        'HC insert_move':   {'best': 943,  'avg':  969, 'time': 0.74},
        'HC reverse_seg':   {'best': 948,  'avg':  981, 'time': 0.80},
        'HC rand-restart':  {'best': 971,  'avg':  989, 'time': 0.85},
    },
    'abz7': {
        'HC single':        {'best': 710,  'avg':  735, 'time': 2.19},
        'HC insert_move':   {'best': 695,  'avg':  720, 'time': 2.46},
        'HC reverse_seg':   {'best': 751,  'avg':  795, 'time': 2.59},
        'HC rand-restart':  {'best': 807,  'avg':  829, 'time': 2.27},
    },
}

instances = ['abz5', 'abz6', 'abz7']
ES_COLOR  = '#4C72B0'   
HC_COLOR  = '#DD8452'   

fig, axes = plt.subplots(2, 3, figsize=(16, 9))
fig.suptitle('(μ+λ) Evolution Strategy  vs  Hill Climbing — ABZ Benchmarks',
             fontsize=14, fontweight='bold', y=1.01)

for col, inst in enumerate(instances):
    opt = OPTIMAL[inst]

    es_labels = list(ES_RESULTS[inst].keys())
    hc_labels = list(HC_RESULTS[inst].keys())
    all_labels = es_labels + hc_labels

    es_best = [ES_RESULTS[inst][k]['best'] for k in es_labels]
    hc_best = [HC_RESULTS[inst][k]['best'] for k in hc_labels]
    es_avg  = [ES_RESULTS[inst][k]['avg']  for k in es_labels]
    hc_avg  = [HC_RESULTS[inst][k]['avg']  for k in hc_labels]

    ax = axes[0, col]
    x  = np.arange(len(all_labels))
    n_es = len(es_labels)
    bars = ax.bar(x[:n_es], es_best, color=ES_COLOR, alpha=0.85, label='ES')
    bars2 = ax.bar(x[n_es:], hc_best, color=HC_COLOR, alpha=0.85, label='HC')
    ax.axhline(opt, color='red', lw=1.5, ls='--', label=f'Optimal ({opt})')
    ax.set_title(f'{inst.upper()} — Best Makespan', fontweight='bold')
    ax.set_xticks(x)
    ax.set_xticklabels(all_labels, rotation=30, ha='right', fontsize=8)
    ax.set_ylabel('Makespan')
    ax.set_ylim(opt * 0.95, max(es_best + hc_best) * 1.03)
    ax.legend(fontsize=8)
    for bar, val in zip(list(bars) + list(bars2), es_best + hc_best):
        gap = (val - opt) / opt * 100
        ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + opt*0.002,
                f'+{gap:.1f}%', ha='center', va='bottom', fontsize=7, color='#333')

    ax = axes[1, col]
    bars3 = ax.bar(x[:n_es], es_avg, color=ES_COLOR, alpha=0.85, label='ES')
    bars4 = ax.bar(x[n_es:], hc_avg, color=HC_COLOR, alpha=0.85, label='HC')
    ax.axhline(opt, color='red', lw=1.5, ls='--', label=f'Optimal ({opt})')
    ax.set_title(f'{inst.upper()} — Average Makespan (20 runs)', fontweight='bold')
    ax.set_xticks(x)
    ax.set_xticklabels(all_labels, rotation=30, ha='right', fontsize=8)
    ax.set_ylabel('Makespan')
    ax.set_ylim(opt * 0.95, max(es_avg + hc_avg) * 1.03)
    ax.legend(fontsize=8)
    for bar, val in zip(list(bars3) + list(bars4), es_avg + hc_avg):
        gap = (val - opt) / opt * 100
        ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + opt*0.002,
                f'+{gap:.1f}%', ha='center', va='bottom', fontsize=7, color='#333')

plt.tight_layout()
plt.show()


fig, axes = plt.subplots(1, 3, figsize=(15, 5))
fig.suptitle('Quality-vs-Speed Trade-off: Best Config per Algorithm',
             fontsize=13, fontweight='bold')

for col, inst in enumerate(instances):
    ax = axes[col]
    opt = OPTIMAL[inst]

    # ES points
    for label, d in ES_RESULTS[inst].items():
        gap  = (d['best'] - opt) / opt * 100
        ax.scatter(d['time'], gap, color=ES_COLOR, s=120, zorder=5, marker='s')
        ax.annotate(label, (d['time'], gap),
                    textcoords='offset points', xytext=(5, 3),
                    fontsize=7.5, color=ES_COLOR)

    # HC points
    for label, d in HC_RESULTS[inst].items():
        gap  = (d['best'] - opt) / opt * 100
        ax.scatter(d['time'], gap, color=HC_COLOR, s=120, zorder=5, marker='o')
        ax.annotate(label, (d['time'], gap),
                    textcoords='offset points', xytext=(5, -7),
                    fontsize=7.5, color=HC_COLOR)

    ax.set_xlabel('Avg time per run (s)', fontsize=10)
    ax.set_ylabel('Gap from optimal (%)', fontsize=10)
    ax.set_title(f'{inst.upper()} (Optimal={opt})', fontweight='bold')
    ax.axhline(0, color='red', lw=1, ls='--', alpha=0.5)
    es_patch = mpatches.Patch(color=ES_COLOR, label='ES (square)')
    hc_patch = mpatches.Patch(color=HC_COLOR, label='HC (circle)')
    ax.legend(handles=[es_patch, hc_patch], fontsize=8)
    ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()


# ── Figure 3: Summary heatmap — avg gap from optimal ──────────────────────
all_algos = [
    'ES es1', 'ES es2', 'ES es3',
    'HC single', 'HC insert_move', 'HC reverse_seg', 'HC rand-restart'
]
avg_data = {
    'abz5': [1309.55, 1308.20, 1297.10, 1269.3, 1262.8, 1290.5, 1288.7],
    'abz6': [1006.20, 1003.40,  995.75,  968.4,  969.3,  981.5,  989.6],
    'abz7': [ 788.80,  780.30,  767.20,  735.2,  720.0,  795.2,  829.9],
}

gap_matrix = np.array([
    [(avg_data[inst][i] - OPTIMAL[inst]) / OPTIMAL[inst] * 100
     for inst in instances]
    for i in range(len(all_algos))
])

fig, ax = plt.subplots(figsize=(8, 6))
im = ax.imshow(gap_matrix, cmap='RdYlGn_r', aspect='auto')
ax.set_xticks(range(len(instances)))
ax.set_xticklabels([i.upper() for i in instances], fontsize=11)
ax.set_yticks(range(len(all_algos)))
ax.set_yticklabels(all_algos, fontsize=10)

ax.axhline(2.5, color='white', lw=3)
ax.text(2.6, 0.9, '— ES —', fontsize=9, color='white', va='center')
ax.text(2.6, 4.5, '— HC —', fontsize=9, color='white', va='center')

for i in range(len(all_algos)):
    for j in range(len(instances)):
        ax.text(j, i, f'{gap_matrix[i, j]:.1f}%',
                ha='center', va='center', fontsize=10, fontweight='bold',
                color='black')

plt.colorbar(im, ax=ax, label='Avg gap from optimal (%)')
ax.set_title('Average Gap from Optimal — ES vs HC (all configs & instances)',
             fontweight='bold', pad=12)
plt.tight_layout()
plt.show()
```

## Conclusion

Aspects:

Search styles - Hill climbing uses a search style more of a form of exploitation, it tries to improve it's current solution. While GA also does abit of exploitation but this happens towards the end, in the beginning it explores differnt solutions trying to achieve a bigger variety.

Optimum - Hill climbing can easily get trapped in a local optimum, while GA is more resistant because it is more diverse. Basically GA has much more data to work with and therefore more alternatives which prevents it from being stuck unlike Hill climbing that does not have as many alternatives.

Convergence speed - Hill climbing converges fast and early which yields faster results but not always does it prioritize quality. GA is usually slower but towards the end it will yield higher Quality, in this area we sacrifice time for quality.

Computational cost - As mentioned above GA has a higher computational cost than Hill climbing.
